```python
import pandas as pd

matches = pd.read_csv('../data/processed/features_v4.csv')

# --------------------------------------------------
# Venue-specific differences
# Home team's last 5 HOME games
# vs Away team's last 5 AWAY games
# --------------------------------------------------

matches['HomeAwayPPGDiff'] = (
    matches['HomeHomePointsLast5'] -
    matches['AwayAwayPointsLast5']
)

matches['HomeAwayGoalsDiff'] = (
    matches['HomeHomeGoalsLast5'] -
    matches['AwayAwayGoalsLast5']
)

# Positive = home team has better defensive record
matches['HomeAwayGoalsAgainstDiff'] = (
    matches['AwayAwayGoalsAgainstLast5'] -
    matches['HomeHomeGoalsAgainstLast5']
)

matches['HomeAwayShotsDiff'] = (
    matches['HomeShotsForLast5'] -
    matches['AwayShotsForLast5']
)

matches['HomeAwayShotsOTDiff'] = (
    matches['HomeShotsOnTargetLast5'] -
    matches['AwayShotsOnTargetLast5']
)
```

```python
matches.to_csv('../data/processed/features_v5.csv', index=False)

print('Saved features_v5.csv')
```
